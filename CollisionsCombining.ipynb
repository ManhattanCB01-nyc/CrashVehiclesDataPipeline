### Incremental Updates to BigQuery Table

To update your BigQuery table daily without re-uploading all historical data, we need to:

1.  **Retrieve the maximum `CRASH DATE`** from your existing `manhattan_filtered_crashes` table in BigQuery.
2.  **Filter your new `crashes_collisions` DataFrame** (representing the latest batch of data) to include only records with a `CRASH DATE` *after* the one currently in BigQuery.
3.  **Append** these new records to the BigQuery table using `if_exists='append'`.

```python
import pandas_gbq

# 1. Retrieve the maximum CRASH DATE from the existing BigQuery table
query = """SELECT MAX(`CRASH DATE`) FROM  stable-liberty-426016-d2.Collisions.manhattan_filtered_crashes"""

try:
    max_date_in_bq_df = pandas_gbq.read_gbq(query, project_id=project_id)
    max_crash_date_in_bq = max_date_in_bq_df.iloc[0, 0]

    if pd.isna(max_crash_date_in_bq):
        print("BigQuery table is empty or 'CRASH DATE' column not found. All data will be uploaded.")
        # If BQ table is empty, treat all current data as new
        new_data_to_upload = crashes_collisions.copy()
    else:
        print(f"Latest crash date in BigQuery: {max_crash_date_in_bq}")
        # 2. Filter new data to include only records after the latest date in BigQuery
        new_data_to_upload = crashes_collisions[crashes_collisions['CRASH DATE'] > max_crash_date_in_bq].copy()

except Exception as e:
    print(f"Error querying BigQuery for max date: {e}")
    print("Proceeding to upload all current data as new to handle potential initial upload or schema issues.")
    new_data_to_upload = crashes_collisions.copy()


if not new_data_to_upload.empty:
    # 3. Append new records to the BigQuery table
    pandas_gbq.to_gbq(
        new_data_to_upload,
        f'{dataset_name}.{table_name}',
        project_id=project_id,
        if_exists='append' # Use 'append' for incremental updates
    )
    print(f"Successfully appended {len(new_data_to_upload)} new records to BigQuery table: stable-liberty-426016-d2.Collisions.manhattan_filtered_crashes")
else:
    print("No new records found to append based on the latest date in BigQuery.")
```

### Incremental Updates for the `all_combined_crashes_data` BigQuery Table

To update your final combined BigQuery table daily, we need to:

1.  **Retrieve the maximum `CRASH_DATE`** from your existing `all_combined_crashes_data` table in BigQuery.
2.  **Filter your new `final_combined_df`** (representing the latest batch of data) to include only records with a `CRASH_DATE` *after* the one currently in BigQuery.
3.  **Append** these new records to the BigQuery table using `if_exists='append'`.

```python
import pandas_gbq

# 1. Retrieve the maximum CRASH DATE from the existing BigQuery table
# Note: Using CRASH_DATE as inferred from previous dataframes. Adjust if BigQuery schema uses 'CRASH DATE' with space.
query = f"""SELECT MAX(`CRASH_DATE`) FROM `{project_id}.{dataset_name}.{final_table_name}`"""

try:
    max_date_in_bq_df_final = pandas_gbq.read_gbq(query, project_id=project_id)
    max_crash_date_in_bq_final = max_date_in_bq_df_final.iloc[0, 0]

    if pd.isna(max_crash_date_in_bq_final):
        print("BigQuery table is empty or 'CRASH_DATE' column not found. All data from current final_combined_df will be uploaded.")
        # If BQ table is empty, treat all current data as new
        new_data_to_upload_final = final_combined_df.copy()
    else:
        print(f"Latest crash date in BigQuery (final table): {max_crash_date_in_bq_final}")
        # 2. Filter new data to include only records after the latest date in BigQuery
        # Assuming 'CRASH_DATE' is the common date column in final_combined_df after merging
        new_data_to_upload_final = final_combined_df[final_combined_df['CRASH_DATE'] > max_crash_date_in_bq_final].copy()

except Exception as e:
    print(f"Error querying BigQuery for max date in final table: {e}")
    print("Proceeding to upload all current final_combined_df data as new to handle potential initial upload or schema issues.")
    new_data_to_upload_final = final_combined_df.copy()


if not new_data_to_upload_final.empty:
    # 3. Append new records to the BigQuery table
    pandas_gbq.to_gbq(
        new_data_to_upload_final,
        f'{dataset_name}.{final_table_name}',
        project_id=project_id,
        if_exists='append' # Use 'append' for incremental updates
    )
    print(f"Successfully appended {len(new_data_to_upload_final)} new records to BigQuery table: {project_id}.{dataset_name}.{final_table_name}")
else:
    print("No new records found to append to the final table based on the latest date in BigQuery.")
```

### Removing Duplicates from `final_combined_df` and Re-uploading to BigQuery

To ensure data integrity, we will remove any duplicate rows from the `final_combined_df` and then upload this deduplicated version to BigQuery, replacing the existing table. This will guarantee that each row in your BigQuery table is unique.

```python
# Identify and remove duplicate rows from final_combined_df
original_rows = final_combined_df.shape[0]
final_combined_df_deduplicated = final_combined_df.drop_duplicates()

duplicate_rows = original_rows - final_combined_df_deduplicated.shape[0]

print(f"Original number of rows: {original_rows}")
print(f"Number of duplicate rows removed: {duplicate_rows}")
print(f"Number of rows after removing duplicates: {final_combined_df_deduplicated.shape[0]}")

display(final_combined_df_deduplicated.head())
```

```python
import pandas_gbq

# Re-upload the deduplicated DataFrame to BigQuery, replacing the existing table
# The table name 'all_combined_crashes_data' and dataset name 'Collisions' are assumed from previous steps.

final_table_name = 'all_combined_crashes_data' # @param {type:"string"}

pandas_gbq.to_gbq(
    final_combined_df_deduplicated,
    f'{dataset_name}.{final_table_name}',
    project_id=project_id,
    if_exists='replace' # Replace the existing table with the deduplicated data
)

print(f"Deduplicated DataFrame successfully uploaded to BigQuery table: {project_id}.{dataset_name}.{final_table_name} (replaced)")
```
